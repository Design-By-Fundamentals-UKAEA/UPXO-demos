```python
import pyvista as pv
import numpy as np

def visualize_3d_array(array, alpha=0.35):
    """Visualize a 3D NumPy array using PyVista ImageData."""
    if array.ndim != 3:
        raise ValueError("`array` must be a 3D NumPy array.")
    if not (0.0 <= alpha <= 1.0):
        raise ValueError("`alpha` must be between 0 and 1.")

    grid = pv.ImageData()
    grid.dimensions = np.array(array.shape) + 1  # cell-centered data
    grid.origin = (0.0, 0.0, 0.0)
    grid.spacing = (1.0, 1.0, 1.0)
    grid.cell_data["values"] = array.flatten(order="F")

    p = pv.Plotter()
    p.add_mesh(grid, scalars="values", opacity=alpha, cmap="viridis")
    p.add_axes()
    p.show_grid()
    p.show()
```

```python
# Volume filled with label 1
shape = (10, 10, 10)  # (z, y, x)
img = np.ones(shape, dtype=np.uint8)

# Half-sphere shell (boundary) with label 2, touching the z=0 boundary
r = 5
thickness = 1
cz, cy, cx = shape[0] // 2, shape[1] // 2, 0  # center on boundary plane x=0

z, y, x = np.ogrid[:shape[0], :shape[1], :shape[2]]
dist = np.sqrt((z - cz) ** 2 + (y - cy) ** 2 + (x - cx) ** 2)

# Shell voxels of the hemisphere inside the domain
half_sphere_boundary = np.abs(dist - r) <= thickness
img[half_sphere_boundary] = 2

# img is the requested 3D labeled image
# label 1: background
# label 2: half-sphere boundary
print("Shape:", img.shape, "dtype:", img.dtype)
print("Unique labels:", np.unique(img))
```

```python
visualize_3d_array(img, alpha=0.8)
```
